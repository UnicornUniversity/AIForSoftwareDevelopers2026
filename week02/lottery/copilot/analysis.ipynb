# Eurojackpot analysis

This notebook reproduces the frequency analysis, heuristic scoring, and ticket suggestion used for the lottery dataset.


```python
import csv
from collections import Counter
from itertools import combinations
from pathlib import Path

csv_path = Path(r'C:\Leonov\AIForSoftwareDevelopers2026\week02\lottery\eurojackpot.csv')
with csv_path.open(encoding='utf-8-sig', newline='') as f:
    rows = list(csv.reader(f, delimiter=';'))
data = [r for r in rows[1:] if len(r) >= 10]
main = Counter()
stars = Counter()
pairs = Counter()
for row in data:
    nums = [int(x.strip()) for x in row[3:10] if x.strip()]
    m, s = nums[:5], nums[5:7]
    main.update(m)
    stars.update(s)
    for a, b in combinations(sorted(m), 2):
        pairs[(a, b)] += 1
```

```python
main_top = sorted(((n, main[n]) for n in range(1, 51)), key=lambda kv: (-kv[1], kv[0]))[:10]
stars_top = sorted(((n, stars[n]) for n in range(1, 13)), key=lambda kv: (-kv[1], kv[0]))[:10]
print("Top main numbers:", main_top[:5])
print("Top stars:", stars_top[:5])
print("Strongest pair:", pairs.most_common(1)[0])
```

```python
# Simple heuristic model: long-term frequency + recent momentum
recent = data[:100]
main_recent = Counter()
stars_recent = Counter()
for row in recent:
    nums = [int(x.strip()) for x in row[3:10] if x.strip()]
    main_recent.update(nums[:5])
    stars_recent.update(nums[5:7])

def main_score(n):
    return main[n] / len(data) + 2 * main_recent[n] / len(recent)

def star_score(n):
    return stars[n] / len(data) + 2 * stars_recent[n] / len(recent)

suggested_main = sorted(range(1, 51), key=lambda n: (-main_score(n), n))[:5]
suggested_stars = sorted(range(1, 13), key=lambda n: (-star_score(n), n))[:2]
print("Suggested main numbers:", suggested_main)
print("Suggested stars:", suggested_stars)
```
